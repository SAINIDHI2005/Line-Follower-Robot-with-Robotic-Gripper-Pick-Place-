```python
import pandas as pd
import numpy as np
import torch

from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import kneighbors_graph
from sklearn.metrics import accuracy_score, classification_report

from torch_geometric.data import Data
from torch_geometric.nn import GCNConv
```

```python
benign_train = pd.read_csv('/content/BenignTraffic.pcap.csv')

ddos_train = pd.read_csv('/content/DDoS-UDP_Flood.pcap.csv')
```

```python
benign_test = pd.read_csv('BenignTraffic1.pcap.csv')

ddos_test = pd.read_csv('DDoS-UDP_Flood1.pcap.csv')
```

```python
benign_train['Label'] = 0
ddos_train['Label'] = 1

benign_test['Label'] = 0
ddos_test['Label'] = 1
```

```python
train_df = pd.concat(
    [benign_train, ddos_train],
    ignore_index=True
)

test_df = pd.concat(
    [benign_test, ddos_test],
    ignore_index=True
)
```

```python
train_df = train_df.sample(10000, random_state=42)

test_df = test_df.sample(3000, random_state=42)
```

```python
X_train = train_df.drop(columns=['Label'])

y_train = train_df['Label']

X_test = test_df.drop(columns=['Label'])

y_test = test_df['Label']
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
A_train = kneighbors_graph(
    X_train,
    n_neighbors=5,
    mode='connectivity',
    include_self=False
)

A_test = kneighbors_graph(
    X_test,
    n_neighbors=5,
    mode='connectivity',
    include_self=False
)
```

```python
edge_index_train = torch.tensor(
    np.array(A_train.nonzero()),
    dtype=torch.long
)

edge_index_test = torch.tensor(
    np.array(A_test.nonzero()),
    dtype=torch.long
)
```

```python
x_train = torch.tensor(X_train, dtype=torch.float)

y_train = torch.tensor(y_train.values, dtype=torch.long)

x_test = torch.tensor(X_test, dtype=torch.float)

y_test = torch.tensor(y_test.values, dtype=torch.long)
```

```python
train_data = Data(
    x=x_train,
    edge_index=edge_index_train,
    y=y_train
)

test_data = Data(
    x=x_test,
    edge_index=edge_index_test,
    y=y_test
)
```

```python
class GCN(torch.nn.Module):

    def __init__(self, input_dim):

        super().__init__()

        self.conv1 = GCNConv(input_dim, 64)

        self.conv2 = GCNConv(64, 32)

        self.fc = torch.nn.Linear(32, 2)

    def forward(self, data):

        x, edge_index = data.x, data.edge_index

        x = self.conv1(x, edge_index)

        x = torch.relu(x)

        x = self.conv2(x, edge_index)

        x = torch.relu(x)

        x = self.fc(x)

        return x
```

```python
model = GCN(input_dim=x_train.shape[1])
```

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.01
)

criterion = torch.nn.CrossEntropyLoss()
```

```python
for epoch in range(50):

    optimizer.zero_grad()

    out = model(train_data)

    loss = criterion(out, train_data.y)

    loss.backward()

    optimizer.step()

    print(f"Epoch {epoch}, Loss: {loss.item()}")
```

```python
model.eval()

with torch.no_grad():

    pred = model(test_data).argmax(dim=1)

accuracy = accuracy_score(
    y_test,
    pred
)

print("Accuracy:", accuracy)
```

```python
print(classification_report(y_test, pred))
```
